# LLM pipeline for screening and scoring open-ended responses

This notebook follows the **question-by-question** workflow:

1. load `question_info.csv`
2. build module summaries from the course folder
3. preprocess responses
4. binary prescreen each response as **insufficient** or **scorable**
5. generate a rubric for one question
6. score that question
7. save outputs
8. move to the next question

The notebook assumes:

- a cleaned response file, such as `Responses_clean.xlsx`
- a `question_info.csv` file
- a `course_folder/` that contains one folder per module, and each module folder contains one or more Jupyter notebooks

#### Build module summaries from the course folder
Defines a helper function that pulls teaching-relevant text from course notebooks.  
It focuses mainly on markdown content, with a light pass over explanatory code comments, so the later summaries reflect the educational content rather than notebook mechanics.

````python

def extract_teaching_text_from_notebook(
    nb_path: Path,
    max_markdown_chars: int = 12000,
    max_code_comment_chars: int = 2000,
) -> Dict[str, str]:
    """
    Extract teaching-relevant content from a notebook.
    """
    nb = nbformat.read(nb_path, as_version=4)

    markdown_parts = []
    code_comment_parts = []

    for cell in nb.cells:
        if cell.cell_type == "markdown":
            text = str(cell.source).strip()
            if text:
                markdown_parts.append(text)

        elif cell.cell_type == "code":
            src = str(cell.source)
            comment_lines = []
            for line in src.splitlines():
                striped = line.strip()
                if striped.startswith("#"):
                    comment_lines.append(striped.lstrip("#").strip())
            if comment_lines:
                code_comment_parts.append("\n".join(comment_lines))

    markdown_text = "\n\n".join(markdown_parts)[:max_markdown_chars]
    code_comment_text = "\n\n".join(code_comment_parts)[:max_code_comment_chars]

    return {
        "markdown_text": markdown_text,
        "code_comment_text": code_comment_text,
    }


def extract_json_block(text: str) -> dict:
    text = text.strip()

    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)

    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass

    match = re.search(r'\{.*\}', text, flags=re.DOTALL)
    if match:
        return json.loads(match.group(0))

    raise ValueError("Could not parse JSON from model output.")
````

Defines helper functions for calling LLM providers and extracting structured JSON output.  
These wrappers standardize model calls so later stages can reuse the same interface.

````python
def get_max_output_tokens(default: int = 6000) -> int:
    try:
        return int(MODEL_CONFIG.get("max_output_tokens", default))
    except Exception:
        return default


def call_openai_json(
    model: str,
    system_prompt: str,
    user_prompt: str,
    max_output_tokens: Optional[int] = None,
) -> dict:
    if openai_client is None:
        raise RuntimeError("OpenAI client is not configured. Check OPENAI_API_KEY in llm-compare.env or .env.")

    response = openai_client.responses.create(
        model=model,
        input=[
            {"role": "system", "content": [{"type": "input_text", "text": system_prompt}]},
            {"role": "user", "content": [{"type": "input_text", "text": user_prompt}]},
        ],
        max_output_tokens=max_output_tokens or get_max_output_tokens(),
    )
    return extract_json_block(response.output_text or "")


def call_bedrock_claude_json(
    model: str,
    system_prompt: str,
    user_prompt: str,
    max_output_tokens: Optional[int] = None,
) -> dict:
    if boto3 is None:
        raise RuntimeError("boto3 is not installed. Run the setup/install cell first.")

    region = os.getenv("AWS_REGION", "us-west-2")
    aws_profile = os.getenv("AWS_PROFILE")

    if aws_profile:
        session = boto3.Session(profile_name=aws_profile, region_name=region)
        client = session.client("bedrock-runtime", region_name=region)
    else:
        client = boto3.client("bedrock-runtime", region_name=region)

    prompt = f"{system_prompt}\n\n{user_prompt}"

    # Some newer Anthropic Claude models on Amazon Bedrock reject the `temperature`
    # parameter in the Converse API (for example: "`temperature` is deprecated for this model.").
    # For scoring, we only need deterministic JSON and max token control, so we omit
    # temperature for Bedrock by default. OpenAI/Gemini wrappers still set temperature separately.
    inference_config = {
        "maxTokens": max_output_tokens or get_max_output_tokens(),
    }

    # Optional escape hatch: set AWS_BEDROCK_SEND_TEMPERATURE=true only if your
    # selected Bedrock model explicitly supports temperature.
    if os.getenv("AWS_BEDROCK_SEND_TEMPERATURE", "false").strip().lower() in {"1", "true", "yes"}:
        inference_config["temperature"] = 0.0

    response = client.converse(
        modelId=model,
        messages=[
            {
                "role": "user",
                "content": [{"text": prompt}],
            }
        ],
        inferenceConfig=inference_config,
    )

    text = response["output"]["message"]["content"][0]["text"]
    return extract_json_block(text)


def call_vertex_gemini_json(
    model: str,
    system_prompt: str,
    user_prompt: str,
    max_output_tokens: Optional[int] = None,
) -> dict:
    if genai is None or genai_types is None:
        raise RuntimeError("google-genai is not installed. Run the setup/install cell first.")

    project = os.getenv("GOOGLE_CLOUD_PROJECT")
    if not project:
        raise RuntimeError("GOOGLE_CLOUD_PROJECT is not set in llm-compare.env or .env.")

    location = os.getenv("GOOGLE_CLOUD_LOCATION", "global")

    client = genai.Client(
        vertexai=True,
        project=project,
        location=location,
        http_options=genai_types.HttpOptions(api_version="v1"),
    )

    response = client.models.generate_content(
        model=model,
        contents=f"{system_prompt}\n\n{user_prompt}",
        config=genai_types.GenerateContentConfig(
            max_output_tokens=max_output_tokens or get_max_output_tokens(),
            temperature=0.0,
        ),
    )

    return extract_json_block(response.text or "")


def call_anthropic_json(
    model: str,
    system_prompt: str,
    user_prompt: str,
    max_output_tokens: Optional[int] = None,
) -> dict:
    """Optional native Anthropic API support, kept for backward compatibility."""
    if anthropic_client is None:
        raise RuntimeError("Native Anthropic client is not configured.")

    response = anthropic_client.messages.create(
        model=model,
        max_tokens=max_output_tokens or get_max_output_tokens(),
        system=system_prompt,
        messages=[{"role": "user", "content": user_prompt}],
    )
    text = "".join(block.text for block in response.content if hasattr(block, "text"))
    return extract_json_block(text)


def call_gemini_json(
    model: str,
    system_prompt: str,
    user_prompt: str,
    max_output_tokens: Optional[int] = None,
) -> dict:
    """Alias for Vertex Gemini, matching the previous notebook provider name."""
    return call_vertex_gemini_json(
        model=model,
        system_prompt=system_prompt,
        user_prompt=user_prompt,
        max_output_tokens=max_output_tokens,
    )


def call_model_json(provider: str, model: str, system_prompt: str, user_prompt: str) -> dict:
    provider = provider.lower().strip()

    if provider == "openai":
        return call_openai_json(model=model, system_prompt=system_prompt, user_prompt=user_prompt)

    if provider in {"aws_bedrock_claude", "bedrock", "bedrock_claude"}:
        return call_bedrock_claude_json(model=model, system_prompt=system_prompt, user_prompt=user_prompt)

    if provider in {"gcp_vertex_gemini", "vertex_gemini", "google_vertex", "google"}:
        return call_vertex_gemini_json(model=model, system_prompt=system_prompt, user_prompt=user_prompt)

    if provider == "anthropic":
        return call_anthropic_json(model=model, system_prompt=system_prompt, user_prompt=user_prompt)

    raise ValueError(f"Unsupported provider: {provider}")
````

Defines the JSON output schemas for notebook-level and module-level summaries.  
These schemas help validate that the model returns the expected fields before the summaries are used downstream.

````python

class NotebookSummarySchema(BaseModel):
    notebook_name: str
    summary: str
    key_concepts: List[str] = Field(default_factory=list)

class ModuleSummarySchema(BaseModel):
    module_id: str
    module_folder: str
    module_summary: str
    learning_goals: List[str] = Field(default_factory=list)
    key_concepts: List[str] = Field(default_factory=list)
````

Builds the prompt for summarizing one teaching notebook.  
The prompt tells the model to focus on educational content and ignore notebook setup details that are not relevant to learner understanding.

````python

def build_notebook_summary_prompt(notebook_name: str, extracted: Dict[str, str]) -> Tuple[str, str]:
    system_prompt = (
        "You are reading teaching materials from a course notebook. "
        "Summarize the educational content only. Ignore notebook mechanics, environment setup, and long technical code details unless they teach concepts. "
        "Return valid JSON only."
    )

    user_prompt = f"""
Notebook name:
{notebook_name}

Markdown content:
{extracted["markdown_text"]}

Code comments:
{extracted["code_comment_text"]}

Task:
Create a concise notebook summary for downstream educational scoring.
Focus on what learners were expected to understand from this notebook.

Return JSON:
{{
  "notebook_name": "{notebook_name}",
  "summary": "2-5 sentence summary",
  "key_concepts": ["concept 1", "concept 2"]
}}
"""
    return system_prompt, user_prompt


def summarize_notebook(nb_path: Path, model_cfg: Dict[str, str]) -> Dict[str, Any]:
    extracted = extract_teaching_text_from_notebook(
        nb_path=nb_path,
        max_markdown_chars=MODEL_CONFIG["max_markdown_chars_per_notebook"],
        max_code_comment_chars=MODEL_CONFIG["max_code_comment_chars_per_notebook"],
    )
    system_prompt, user_prompt = build_notebook_summary_prompt(nb_path.name, extracted)
    result = call_model_json(
        provider=model_cfg["provider"],
        model=model_cfg["model"],
        system_prompt=system_prompt,
        user_prompt=user_prompt,
    )
    return NotebookSummarySchema(**result).model_dump()
````

Builds the prompt for combining several notebook summaries into a single module-level summary.  
This creates the teaching context that will later help with screening and scoring responses.

````python

def build_module_summary_prompt(
    module_id: str,
    module_folder: str,
    notebook_summaries: List[Dict[str, Any]],
) -> Tuple[str, str]:
    system_prompt = (
        "You are building a module-level summary from notebook summaries. "
        "Focus on the educational content that a learner should take away from the module. "
        "Return valid JSON only."
    )

    user_prompt = f"""
Module ID: {module_id}
Module folder: {module_folder}

Notebook summaries:
{json.dumps(notebook_summaries, ensure_ascii=False, indent=2)}

Task:
Combine these notebook summaries into one concise module summary.
Make the summary useful for judging whether a learner response is scorable and for building a scoring rubric.

Return JSON:
{{
  "module_id": "{module_id}",
  "module_folder": "{module_folder}",
  "module_summary": "3-6 sentence summary",
  "learning_goals": ["goal 1", "goal 2"],
  "key_concepts": ["concept 1", "concept 2", "concept 3"]
}}
"""
    return system_prompt, user_prompt


def summarize_module_folder(
    module_id: str,
    module_folder: str,
    course_dir: Path,
    model_cfg: Dict[str, str],
) -> Dict[str, Any]:
    module_dir = course_dir / module_folder
    assert module_dir.exists(), f"Module folder does not exist: {module_dir}"

    notebook_paths = sorted(module_dir.rglob("*.ipynb"))
    assert notebook_paths, f"No notebooks found in module folder: {module_dir}"

    notebook_summaries = []
    for nb_path in notebook_paths:
        notebook_summaries.append(summarize_notebook(nb_path, model_cfg=model_cfg))

    system_prompt, user_prompt = build_module_summary_prompt(
        module_id=module_id,
        module_folder=module_folder,
        notebook_summaries=notebook_summaries,
    )
    module_summary = call_model_json(
        provider=model_cfg["provider"],
        model=model_cfg["model"],
        system_prompt=system_prompt,
        user_prompt=user_prompt,
    )
    module_summary = ModuleSummarySchema(**module_summary).model_dump()
    module_summary["notebook_summaries"] = notebook_summaries
    return module_summary
````

Creates or loads module summaries for the course materials.  
If a saved summary file already exists and overwrite is turned off, it reuses that file; otherwise, it summarizes the notebooks and saves a new module summary file.

````python

def build_or_load_module_summaries(
    question_info: Dict[str, Dict[str, Any]],
    course_dir: Path,
    out_dir: Path,
    model_cfg: Dict[str, str],
    overwrite: bool = False,
) -> Dict[str, Dict[str, Any]]:
    out_file = out_dir / "module_summaries.json"

    if out_file.exists() and not overwrite:
        with open(out_file, "r", encoding="utf-8") as f:
            module_summaries = json.load(f)
        print("Loaded existing module summaries from:", out_file.resolve())
        return module_summaries

    unique_modules = {}
    for meta in question_info.values():
        unique_modules[meta["module_id"]] = meta["module_folder"]

    module_summaries = {}
    for module_id, module_folder in unique_modules.items():
        print(f"Summarizing module {module_id} from folder '{module_folder}' ...")
        module_summaries[module_id] = summarize_module_folder(
            module_id=module_id,
            module_folder=module_folder,
            course_dir=course_dir,
            model_cfg=model_cfg,
        )

    with open(out_file, "w", encoding="utf-8") as f:
        json.dump(module_summaries, f, ensure_ascii=False, indent=2)

    print("Saved module summaries to:", out_file.resolve())
    return module_summaries
````

Runs the module-summary workflow and stores the resulting summaries in `module_summaries`.  
These summaries are later used as context for response screening and scoring.

````python

module_summaries = build_or_load_module_summaries(
    question_info=QUESTION_INFO,
    course_dir=COURSE_DIR,
    out_dir=MODULE_SUMMARY_DIR,
    model_cfg=MODEL_CONFIG["module_summary_model"],
    overwrite=MODEL_CONFIG["overwrite_module_summaries"],
)

list(module_summaries.keys())
````

## Stage 2. Preprocess responses


#### reshape dataset to long format
Defines a function to reshape the response dataset from wide format to long format.  
After this step, each row represents one learner's response for one question at one timepoint (`pre` or `post`), which makes later screening and scoring much easier.

````python

def reshape_to_long(df: pd.DataFrame, question_info: Dict[str, Dict[str, Any]]) -> pd.DataFrame:
    rows = []

    id_col = "ID"
    assert id_col in df.columns, "Expected an 'ID' column in the response dataset."

    for _, row in df.iterrows():
        respondent_id = row[id_col]

        for qid, meta in question_info.items():
            for timepoint, col in [("pre", meta["pre_col"]), ("post", meta["post_col"])]:
                rows.append({
                    "ID": respondent_id,
                    "question_id": qid,
                    "topic": meta["topic"],
                    "timepoint": timepoint,
                    "question_text": meta["question_text"],
                    "module_id": meta["module_id"],
                    "module_folder": meta["module_folder"],
                    "response_raw": row.get(col, None),
                })

    return pd.DataFrame(rows)


def normalize_text(text: Any) -> str:
    if pd.isna(text):
        return ""

    text = str(text)

    text = (
        text.replace("“", '"')
            .replace("”", '"')
            .replace("‘", "'")
            .replace("’", "'")
    )

    text = text.strip()

    if len(text) >= 2:
        if (text.startswith('"') and text.endswith('"')) or (text.startswith("'") and text.endswith("'")):
            text = text[1:-1].strip()

    text = re.sub(r"\s+", " ", text)
    return text
````

Applies the reshape function, creates cleaned response text, and computes basic text features such as character count and word count.  
These features are useful for both rule-based screening and general quality checks.

````python

long_df = reshape_to_long(raw_df, QUESTION_INFO)
long_df["response_text_raw"] = long_df["response_raw"]
long_df["response_clean"] = long_df["response_raw"].apply(normalize_text)
long_df["char_count"] = long_df["response_clean"].str.len()
long_df["word_count"] = long_df["response_clean"].fillna("").str.split().apply(len)

print("Long shape:", long_df.shape)
long_df.head(10)
````

## Stage 3. Check Response Completeness

````python
# section 1: Known insufficient resposnes
# LLM based insufficient responses
## check if LLM screen the insufficient responses filtered after stage 1
````

Defines a set of exact low-information responses, such as blank answers or very non-informative text.  
These rules provide a quick first-pass filter before model-based screening.

````python

LOW_INFO_EXACT = {
    "",
    "n/a",
    "na",
    "none",
    "nothing",
    "not sure",
    "i'm not sure",
    "im not sure",
    "unsure",
    "idk",
    "i don't know",
    "i dont know",
    "don't know",
    "dont know",
    "unknown",
    "no idea",
    "not applicable",
}

LOW_INFO_PATTERNS = [
    r"^n/?a\.?$",
    r"^idk\.?$",
    r"^not sure\.?$",
    r"^i('?| a)?m not sure\.?$",
    r"^i do not know\.?$",
    r"^i don'?t know\.?$",
    r"^dont know\.?$",
    r"^no idea\.?$",
    r"^nothing\.?$",
    r"^none\.?$",
    r"^unknown\.?$",
]

def is_punctuation_only(text: str) -> bool:
    if pd.isna(text):
        return True
    text = str(text).strip()
    if text == "":
        return True
    return re.sub(r"[\W_]+", "", text) == ""

def is_obvious_insufficient(text: str) -> bool:
    if pd.isna(text):
        return True

    x = str(text).strip().lower()
    if x in LOW_INFO_EXACT:
        return True
    if is_punctuation_only(x):
        return True
    return any(re.match(pattern, x) for pattern in LOW_INFO_PATTERNS)
````

Defines the structured schema for the screening stage.  
Each response will later receive:
- a row identifier
- a screening label (`insufficient` or `scorable`)


````python

class BatchScreenItem(BaseModel):
    row_id: str
    label: Literal["insufficient", "scorable"]

class BatchScreenResponse(BaseModel):
    results: List[BatchScreenItem]
    
````

Builds the prompt used for batch response screening.  
The model uses the question text and module summary as context to decide whether each response contains enough meaningful content to be scored.

````python

def build_batch_screen_prompt(
    question_text: str,
    module_summary: dict,
    batch_records: List[dict],
) -> Tuple[str, str]:
    system_prompt = (
        "You are doing a binary prescreen for open-ended educational responses. "
        "Your only job is to decide whether each response is 'insufficient' or 'scorable'. "
        "Use the question and module summary as context. "
        "Short responses can still be scorable if they express a meaningful relevant idea. "
        "Mark a response as insufficient only when there is not enough meaningful content to judge it. "
        "Return valid JSON only."
    )

    return_format = """
Return JSON:
{
  "results": [
    {
      "row_id": "123",
      "label": "scorable"
    }
  ]
}
"""

    user_prompt = f"""
Question:
{question_text}

Module summary:
{json.dumps(module_summary, ensure_ascii=False, separators=(",", ":"))}

Responses to classify:
{json.dumps(batch_records, ensure_ascii=False, separators=(",", ":"))}

Task:
Classify every response as either:
- "insufficient"
- "scorable"

Decision rule:
- Use "insufficient" only when the response is blank, non-substantive, irrelevant, or too vague to judge.
- Use "scorable" whenever the response contains any meaningful idea relevant to the question, even if it is short or weak.

{return_format}
"""
    return system_prompt, user_prompt
    
````

Screens one batch of learner responses using the selected model.  
It prepares the batch records, sends them to the model, validates the structured output, and returns screening results keyed by row.

````python

def screen_batch_of_responses(
    provider: str,
    model: str,
    question_text: str,
    module_summary: dict,
    batch_df: pd.DataFrame,
) -> Dict[str, Dict[str, Any]]:
    batch_records = [
        {"row_id": str(idx), "response_clean": row["response_clean"]}
        for idx, row in batch_df.iterrows()
    ]

    system_prompt, user_prompt = build_batch_screen_prompt(
        question_text=question_text,
        module_summary=module_summary,
        batch_records=batch_records,
    )

    raw = call_model_json(
        provider=provider,
        model=model,
        system_prompt=system_prompt,
        user_prompt=user_prompt,
    )
    parsed = BatchScreenResponse(**raw)

    result = {}
    for item in parsed.results:
        result[item.row_id] = item.model_dump()
    return result
    
````

Defines batching utilities and the full prescreening workflow for one question.  
This step combines rule-based checks and LLM-based screening, then assigns the final screening label used to decide which responses move on to rubric-based scoring.

````python

def iter_question_batches(df_q_valid: pd.DataFrame, batch_size: int):
    indices = df_q_valid.index.to_list()
    for start in range(0, len(indices), batch_size):
        yield indices[start:start + batch_size]


def prescreen_question_dataframe(
    df_q: pd.DataFrame,
    model_cfg: Dict[str, str],
    module_summary: dict,
    batch_size: int = 40,
    sleep_sec: float = 0.0,
) -> pd.DataFrame:
    df_q = df_q.copy()

    df_q["rule_based_insufficient"] = df_q["response_clean"].apply(is_obvious_insufficient)
    df_q["screen_label"] = None

    # Rule-based obvious insufficient responses do not need an LLM screen.
    df_q.loc[df_q["rule_based_insufficient"], "screen_label"] = "insufficient"

    # LLM screen on the rest.
    to_screen = df_q.loc[~df_q["rule_based_insufficient"]].copy()
    if len(to_screen) == 0:
        return df_q

    for idx_batch in tqdm(list(iter_question_batches(to_screen, batch_size)), desc="LLM prescreen"):
        batch_df = df_q.loc[idx_batch]
        result_map = screen_batch_of_responses(
            provider=model_cfg["provider"],
            model=model_cfg["model"],
            question_text=str(df_q["question_text"].iloc[0]),
            module_summary=module_summary,
            batch_df=batch_df,
        )

        for idx in idx_batch:
            key = str(idx)

            if key not in result_map:
                print(f"WARNING: Missing prescreen result for row index {idx}. Retrying this row individually...")
        
                try:
                    retry_map = screen_batch_of_responses(
                        provider=model_cfg["provider"],
                        model=model_cfg["model"],
                        question_text=str(df_q["question_text"].iloc[0]),
                        module_summary=module_summary,
                        batch_df=df_q.loc[[idx]],
                    )
        
                    if key in retry_map:
                        item = retry_map[key]
                        df_q.at[idx, "screen_label"] = item["label"]
                    else:
                        print(f"WARNING: Retry also missing row {idx}. Defaulting to scorable.")
                        df_q.at[idx, "screen_label"] = "scorable"
        
                except Exception as e:
                    print(f"WARNING: Retry failed for row {idx}: {e}. Defaulting to scorable.")
                    df_q.at[idx, "screen_label"] = "scorable"
        
            else:
                item = result_map[key]
                df_q.at[idx, "screen_label"] = item["label"]

        if sleep_sec:
            time.sleep(sleep_sec)

    return df_q
    
````

## Stage 4. Rubric generation


Defines the rubric schema used for question-specific scoring.  
The rubric is built on a 1–3 scale, where responses are classified into performance levels such as beginner, intermediate, and advanced.

````python

class RubricLevel(BaseModel):
    level: int
    label: str
    description: str
    anchor_examples: List[str] = Field(default_factory=list)

class RubricSchema(BaseModel):
    question_id: str
    question_text: str
    scale_min: int
    scale_max: int
    general_scoring_principles: List[str]
    levels: List[RubricLevel]
    insufficient_response_rule: str
````

Defines a helper function to sample pilot responses from the scorable responses only.  
These sampled responses help the model create a more realistic and question-specific scoring rubric.

````python

def sample_pilot_responses(
    df_q: pd.DataFrame,
    sample_prop: float = 1.0,
    min_n: int = 0,
    max_n: int = 10000,
    random_state: int = 42,
) -> List[str]:
    """
    Sample pilot responses for rubric generation from scorable responses only.

    Rules:
    - use 100% of scorable responses by default
    - enforce a minimum sample size
    - enforce a maximum sample size
    - remove empty and duplicate responses
    """
    valid = df_q.loc[df_q["screen_label"] == "scorable", "response_clean"].dropna()
    valid = valid[valid.str.len() > 0].drop_duplicates()

    if len(valid) == 0:
        return []

    n_valid = len(valid)
    n_sample = int(round(n_valid * sample_prop))
    n_sample = max(n_sample, min_n)
    n_sample = min(n_sample, max_n)
    n_sample = min(n_sample, n_valid)

    print(f"Total scorable responses: {n_valid}")
    print(f"Pilot sample size for rubric generation: {n_sample}")

    return valid.sample(n=n_sample, random_state=random_state).tolist()


def build_rubric_prompt(
    qid: str,
    meta: Dict[str, Any],
    module_summary: dict,
    pilot_responses: List[str],
) -> Tuple[str, str]:
    system_prompt = (
        "You are an expert educational assessment assistant. "
        "Create a concise scoring rubric for one open-ended question. "
        "The rubric must use a 1-3 scale: "
        "1 = Beginner: minimal understanding OR insufficient response (e.g., blank, irrelevant, or 'I don't know')\n"
        "2 = Intermediate: partial or basic understanding with some relevant detail\n"
        "3 = Advanced: clear, accurate, and specific understanding\n\n"
        "Treat blank or insufficient responses as score 1. "
        "Return valid JSON only."
    )

    user_prompt = f"""
Question ID: {qid}

Question:
{meta["question_text"]}

Topic:
{meta["topic"]}

Module summary:
{json.dumps(module_summary, ensure_ascii=False, indent=2)}

Sample scorable learner responses:
{json.dumps(pilot_responses, ensure_ascii=False, indent=2)}

Task:
Create a rubric that differentiates beginner, intermediate, and advanced answers for this question.
Make the rubric specific enough for consistent scoring.

Return JSON:
{{
  "question_id": "{qid}",
  "question_text": "{meta["question_text"]}",
  "scale_min": 1,
  "scale_max": 3,
  "general_scoring_principles": [
    "principle 1",
    "principle 2"
  ],
  "levels": [
    {{
      "level": 1,
      "label": "Beginner",
      "description": "description",
      "anchor_examples": ["example 1", "example 2"]
    }},
    {{
      "level": 2,
      "label": "Intermediate",
      "description": "description",
      "anchor_examples": ["example 1", "example 2"]
    }},
    {{
      "level": 3,
      "label": "Advanced",
      "description": "description",
      "anchor_examples": ["example 1", "example 2"]
    }}
  ],
  "insufficient_response_rule": "How to handle blank, irrelevant, or non-substantive responses"
}}
"""
    return system_prompt, user_prompt
    
````

Generates a rubric for one question using:
- the question metadata
- the linked module summary
- a sample of scorable responses

The returned rubric is then validated and saved for reuse.

````python

def generate_rubric_for_question(
    qid: str,
    df_q: pd.DataFrame,
    meta: Dict[str, Any],
    module_summary: dict,
    model_cfg: Dict[str, str],
) -> Dict[str, Any]:
    pilot_responses = sample_pilot_responses(
        df_q,
        sample_prop=1.0,
        min_n=0,
        max_n=1000,
        random_state=42,
    )

    if len(pilot_responses) == 0:
        raise ValueError(f"No scorable pilot responses available for {qid}")

    system_prompt, user_prompt = build_rubric_prompt(
        qid=qid,
        meta=meta,
        module_summary=module_summary,
        pilot_responses=pilot_responses,
    )

    rubric = call_model_json(
        provider=model_cfg["provider"],
        model=model_cfg["model"],
        system_prompt=system_prompt,
        user_prompt=user_prompt,
    )
    return RubricSchema(**rubric).model_dump()
    
````

## Stage 5. Scoring


Defines the structured output schema for the scoring stage.  
Each scored response includes only:
- the row identifier
- a score from 1 to 3


````python

class BatchScoreItem(BaseModel):
    row_id: str
    score: int = Field(ge=1, le=3)

class BatchScoreResponse(BaseModel):
    results: List[BatchScoreItem]
````

Builds the prompt for rubric-based batch scoring.  
The model is instructed to score each response using only the provided question, module summary, and rubric.

````python

def build_batch_scoring_prompt(
    question_text: str,
    module_summary: dict,
    rubric: dict,
    batch_records: List[dict],
) -> Tuple[str, str]:
    system_prompt = (
        "You are a careful scorer of open-ended educational responses. "
        "Use only the provided question, module summary, and rubric. "
        "Score each response independently. "
        "Do not compare learners to each other, do not grade on a curve, and do not use outside assumptions. "
        "Return valid JSON only."
    )

    user_prompt = f"""
Question:
{question_text}

Module summary:
{json.dumps(module_summary, ensure_ascii=False, separators=(",", ":"))}

Rubric:
{json.dumps(rubric, ensure_ascii=False, separators=(",", ":"))}

Responses to score:
{json.dumps(batch_records, ensure_ascii=False, separators=(",", ":"))}

Task:
Score EVERY response independently using the rubric only.

Rules:
1. Return exactly one result for each row_id shown above.
2. Keep the same row_id values exactly as provided.
3. Do not omit any row_id.
4. Focus on relevance, conceptual accuracy, and completeness.
5. Return JSON only, with no extra text.

Return JSON:
{{
  "results": [
    {{
      "row_id": "123",
      "score": 1
    }}
  ]
}}
"""
    return system_prompt, user_prompt
````

Scores one batch of responses with a selected model.  
It sends the batch to the model, validates the returned structure, and stores the score-related outputs for each row.

````python

def score_batch_of_responses(
    provider: str,
    model: str,
    question_text: str,
    module_summary: dict,
    rubric: dict,
    batch_df: pd.DataFrame,
) -> Dict[str, Dict[str, Any]]:
    batch_records = [
        {"row_id": str(idx), "response_clean": row["response_clean"]}
        for idx, row in batch_df.iterrows()
    ]

    system_prompt, user_prompt = build_batch_scoring_prompt(
        question_text=question_text,
        module_summary=module_summary,
        rubric=rubric,
        batch_records=batch_records,
    )

    raw = call_model_json(
        provider=provider,
        model=model,
        system_prompt=system_prompt,
        user_prompt=user_prompt,
    )
    parsed = BatchScoreResponse(**raw)

    result = {}
    for item in parsed.results:
        result[item.row_id] = item.model_dump()
    return result
````

Defines helper functions for combining multiple model scores.  
These functions calculate:
- the majority-vote final score
- the agreement rate across models

This is useful when using more than one model for scoring.

````python

def majority_vote(scores: List[Optional[int]]) -> Optional[int]:
    valid_scores = [s for s in scores if s is not None]
    if not valid_scores:
        return None
    counts = Counter(valid_scores)
    max_count = max(counts.values())
    winners = [k for k, v in counts.items() if v == max_count]
    return int(median(valid_scores))

def agreement_rate(scores: List[Optional[int]]) -> float:
    valid_scores = [s for s in scores if s is not None]
    if not valid_scores:
        return 0.0
    mv = majority_vote(valid_scores)
    return sum(s == mv for s in valid_scores) / len(valid_scores)


def apply_insufficient_score_rule(df: pd.DataFrame) -> pd.DataFrame:
    """
    Ensure responses identified as insufficient are coded as score 1.

    Notes:
    - Insufficient responses are not sent to the LLM scoring models, so model-specific
      score columns can remain missing.
    - final_score is set to 1 so these rows are included in score summaries.
    - final_label is set to "Beginner" because score 1 represents Beginner/insufficient
      according to the rubric. The original insufficient status is still preserved in
      screen_label for separate tracking.
    - agreement remains missing because there was no multi-model scoring agreement to compute.
    """
    df = df.copy()

    if "screen_label" not in df.columns:
        return df

    insufficient_mask = (
        df["screen_label"]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("insufficient")
    )

    if insufficient_mask.any():
        df.loc[insufficient_mask, "final_score"] = 1
        df.loc[insufficient_mask, "final_label"] = "Beginner"
        df.loc[insufficient_mask, "agreement"] = None
        df.loc[insufficient_mask, "review_flag"] = False

    return df

````

Adds retry and fallback logic for batch scoring.  
If a scoring call fails because of a temporary API or parsing issue, this function retries before giving up.

````python

def score_batch_with_fallback(
    provider: str,
    model: str,
    question_text: str,
    module_summary: dict,
    rubric: dict,
    batch_df: pd.DataFrame,
    max_retries: int = 1,
):
    last_error = None

    for _ in range(max_retries + 1):
        try:
            results = score_batch_of_responses(
                provider=provider,
                model=model,
                question_text=question_text,
                module_summary=module_summary,
                rubric=rubric,
                batch_df=batch_df,
            )
            return results, {}
        except Exception as e:
            last_error = e

    if len(batch_df) <= 1:
        idx = str(batch_df.index[0])
        return {}, {idx: str(last_error)}

    mid = len(batch_df) // 2
    left_df = batch_df.iloc[:mid]
    right_df = batch_df.iloc[mid:]

    left_results, left_errors = score_batch_with_fallback(
        provider=provider,
        model=model,
        question_text=question_text,
        module_summary=module_summary,
        rubric=rubric,
        batch_df=left_df,
        max_retries=max_retries,
    )
    right_results, right_errors = score_batch_with_fallback(
        provider=provider,
        model=model,
        question_text=question_text,
        module_summary=module_summary,
        rubric=rubric,
        batch_df=right_df,
        max_retries=max_retries,
    )

    return {**left_results, **right_results}, {**left_errors, **right_errors}
````

Runs the full scoring workflow for one question.  
It processes responses in batches, scores them with one or more models, combines the results, calculates agreement, and flags low-agreement cases for possible human review.

````python

def score_question_dataframe(
    df_q: pd.DataFrame,
    module_summary: dict,
    rubric: dict,
    scoring_models: List[Dict[str, str]],
    batch_size: int = 50,
    agreement_threshold: float = 0.60,
    sleep_sec: float = 0.0,
    max_workers: int = 4,
    fallback_max_retries: int = 1,
) -> pd.DataFrame:
    df_q = df_q.copy()

    # Initialize only the score columns needed for majority voting.
    for model_cfg in scoring_models:
        model_name = model_cfg["name"]
        df_q[f"{model_name}_score"] = None

    scorable_df = df_q.loc[df_q["screen_label"] == "scorable"].copy()

    for model_cfg in scoring_models:
        model_name = model_cfg["name"]
        print(f"Scoring with {model_name} ...")

        futures = []
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            for idx_batch in iter_question_batches(scorable_df, batch_size):
                batch_df = scorable_df.loc[idx_batch]
                futures.append(
                    executor.submit(
                        score_batch_with_fallback,
                        provider=model_cfg["provider"],
                        model=model_cfg["model"],
                        question_text=str(df_q["question_text"].iloc[0]),
                        module_summary=module_summary,
                        rubric=rubric,
                        batch_df=batch_df,
                        max_retries=fallback_max_retries,
                    )
                )

            for fut in tqdm(as_completed(futures), total=len(futures), desc=f"{model_name} batches"):
                results_map, errors_map = fut.result()

                for idx_str, item in results_map.items():
                    idx = int(idx_str)
                    df_q.at[idx, f"{model_name}_score"] = item["score"]

                for idx_str, err in errors_map.items():
                    idx = int(idx_str)
                    df_q.at[idx, f"{model_name}_score"] = None
                    print(f"WARNING: {model_name} failed for row {idx}: {err}")

        if sleep_sec:
            time.sleep(sleep_sec)

    score_cols = [f'{m["name"]}_score' for m in scoring_models]

    df_q["final_score"] = df_q[score_cols].apply(
        lambda row: majority_vote([v for v in row.tolist() if pd.notna(v)]),
        axis=1,
    )
    df_q["agreement"] = df_q[score_cols].apply(
        lambda row: agreement_rate([v for v in row.tolist() if pd.notna(v)]),
        axis=1,
    )
    df_q["review_flag"] = (
        (df_q["screen_label"] == "scorable") &
        (df_q["agreement"] < agreement_threshold)
    )

    # Final label for convenience.
    score_to_label = {1: "Beginner", 2: "Intermediate", 3: "Advanced"}
    df_q["final_label"] = df_q["final_score"].map(score_to_label)

    # Insufficient responses are not sent to the LLM scoring models, but they
    # should still be included in the final numeric score as 1.
    df_q = apply_insufficient_score_rule(df_q)

    return df_q
````

#### One-question end-to-end workflow
Defines the end-to-end workflow for a single question.  
For that question, it:
1. selects the relevant responses  
2. prescreens them  
3. generates or loads a rubric  
4. scores the scorable responses  
5. saves the outputs

````python

def run_one_question_pipeline(
    qid: str,
    long_df: pd.DataFrame,
    question_info: Dict[str, Dict[str, Any]],
    module_summaries: Dict[str, Dict[str, Any]],
    model_config: Dict[str, Any],
) -> Tuple[Dict[str, Any], pd.DataFrame]:
    meta = question_info[qid]
    module_summary = module_summaries[meta["module_id"]]
    overwrite = model_config["overwrite_question_outputs"]

    screen_path = SCREEN_DIR / f"screened_{qid}.csv"
    rubric_path = RUBRIC_DIR / f"rubric_{qid}.json"
    score_path = RESULTS_DIR / f"scored_{qid}.csv"

    df_q = long_df[long_df["question_id"] == qid].copy()

    # 1. Prescreen
    if screen_path.exists() and not overwrite:
        screened_q = pd.read_csv(screen_path)
        print(f"Loaded existing prescreen output for {qid}")
    else:
        print(f"Prescreening {qid} ...")
        screened_q = prescreen_question_dataframe(
            df_q=df_q,
            model_cfg=model_config["screen_model"],
            module_summary=module_summary,
            batch_size=model_config["screen_batch_size"],
            sleep_sec=model_config["sleep_between_calls_sec"],
        )
        screened_q.to_csv(screen_path, index=False)
        print("Saved:", screen_path.resolve())

    # 2. Rubric
    if rubric_path.exists() and not overwrite:
        with open(rubric_path, "r", encoding="utf-8") as f:
            rubric = json.load(f)
        print(f"Loaded existing rubric for {qid}")
    else:
        print(f"Generating rubric for {qid} ...")
        rubric = generate_rubric_for_question(
            qid=qid,
            df_q=screened_q,
            meta=meta,
            module_summary=module_summary,
            model_cfg=model_config["rubric_model"],
        )
        with open(rubric_path, "w", encoding="utf-8") as f:
            json.dump(rubric, f, ensure_ascii=False, indent=2)
        print("Saved:", rubric_path.resolve())

    # 3. Score
    if score_path.exists() and not overwrite:
        scored_q = pd.read_csv(score_path)
        print(f"Loaded existing scores for {qid}")
    else:
        print(f"Scoring {qid} ...")
        scored_q = score_question_dataframe(
            df_q=screened_q,
            module_summary=module_summary,
            rubric=rubric,
            scoring_models=model_config["scoring_models"],
            batch_size=model_config["scoring_batch_size"],
            agreement_threshold=model_config["agreement_threshold"],
            sleep_sec=model_config["sleep_between_calls_sec"],
            max_workers=model_config["max_workers"],
            fallback_max_retries=model_config["fallback_max_retries"],
        )
        scored_q.to_csv(score_path, index=False)
        print("Saved:", score_path.resolve())

    # Apply the insufficient-response scoring rule even when existing score files
    # are loaded, so old outputs with missing final_score values are repaired.
    scored_q = apply_insufficient_score_rule(scored_q)
    scored_q.to_csv(score_path, index=False)

    return rubric, scored_q
````

Runs the full question-by-question pipeline across all questions in `QUESTION_INFO`.  
It stores each question's rubric and scored output, then combines everything into one final dataset.

````python

all_scored = []
rubrics = {}

for qid in QUESTION_INFO.keys():
    print(f"\n===== Running {qid} =====")
    rubric, scored_q = run_one_question_pipeline(
        qid=qid,
        long_df=long_df,
        question_info=QUESTION_INFO,
        module_summaries=module_summaries,
        model_config=MODEL_CONFIG,
    )
    rubrics[qid] = rubric
    all_scored.append(scored_q)

scored_all = pd.concat(all_scored, ignore_index=True)
scored_all = apply_insufficient_score_rule(scored_all)
scored_all.to_csv(RESULTS_DIR / "scored_all_questions.csv", index=False)

print("Done.")
print("Combined output:", (RESULTS_DIR / "scored_all_questions.csv").resolve())
````

## Stage 6. Review and QC


#### Export flagged responses for human review
This creates/updates output/review/flagged_for_review.csv.
You can manually enter human-review results directly into this same file.

````python
flagged_path = REVIEW_DIR / "flagged_for_review.csv"

key_cols = ["ID", "question_id", "timepoint"]

human_review_cols = [
    "human_final_score",   # Enter 1, 2, or 3 after human review
    "human_review_note",   # Optional notes explaining the human decision
    "human_reviewer",      # Optional reviewer name/initials
    "human_review_date"    # Optional review date
]

# Select rows currently flagged for review
flagged_new = scored_all[scored_all["review_flag"] == True].copy()

# Add blank human-review columns to the newly generated flagged file
for col in human_review_cols:
    if col not in flagged_new.columns:
        flagged_new[col] = pd.NA

# Preserve existing human-review results if flagged_for_review.csv already exists
if flagged_path.exists():
    flagged_existing = pd.read_csv(flagged_path)

    # Make sure key columns exist in the existing review file
    missing_keys = [col for col in key_cols if col not in flagged_existing.columns]
    if missing_keys:
        print("Existing flagged_for_review.csv is missing key columns:", missing_keys)
        print("The file will be regenerated without preserving previous human-review results.")
    else:
        # Convert key columns to string to avoid merge issues
        for col in key_cols:
            flagged_new[col] = flagged_new[col].astype(str)
            flagged_existing[col] = flagged_existing[col].astype(str)

        # Keep only existing human-review columns that are actually present
        existing_review_cols = [
            col for col in human_review_cols if col in flagged_existing.columns
        ]

        if existing_review_cols:
            existing_human_review = flagged_existing[
                key_cols + existing_review_cols
            ].copy()

            # Remove blank human-review columns from new flagged file before merging
            flagged_new = flagged_new.drop(
                columns=[col for col in human_review_cols if col in flagged_new.columns],
                errors="ignore"
            )

            # Merge previous human-review results back into the new flagged file
            flagged_new = flagged_new.merge(
                existing_human_review,
                on=key_cols,
                how="left"
            )

            # Re-add any missing human-review columns
            for col in human_review_cols:
                if col not in flagged_new.columns:
                    flagged_new[col] = pd.NA

# Save flagged file
flagged_new.to_csv(flagged_path, index=False)

print("Flagged rows:", len(flagged_new))
print("Saved:", flagged_path.resolve())

flagged_new.head()
````

#### Human review and QC merge (optional)
Merge human-review results from the same flagged_for_review.csv file
If human_final_score is filled in, it replaces the original final_score.
If human_final_score is blank, the original final_score is kept.

````python
flagged_path = REVIEW_DIR / "flagged_for_review.csv"
final_output_path = RESULTS_DIR / "scored_all_questions_final.csv"

key_cols = ["ID", "question_id", "timepoint"]

human_review_cols = [
    "human_final_score",
    "human_review_note",
    "human_reviewer",
    "human_review_date"
]

score_to_label = {
    1: "Beginner",
    2: "Intermediate",
    3: "Advanced"
}

final_df = scored_all.copy()

# Default: keep original model/consensus scores
final_df["final_score_after_qc"] = final_df["final_score"]
final_df["final_label_after_qc"] = final_df["final_label"]
final_df["human_review_completed"] = False

if flagged_path.exists():
    review_df = pd.read_csv(flagged_path)

    # Check required columns
    missing_keys = [col for col in key_cols if col not in review_df.columns]
    if missing_keys:
        print("flagged_for_review.csv is missing key columns:", missing_keys)
        print("Original results kept.")
    elif "human_final_score" not in review_df.columns:
        print("No human_final_score column found in flagged_for_review.csv.")
        print("Original results kept.")
    else:
        # Convert key columns to string to avoid merge issues
        for col in key_cols:
            final_df[col] = final_df[col].astype(str)
            review_df[col] = review_df[col].astype(str)

        # Clean human_final_score
        # This allows entries like 1, 2, 3, "1: Beginner", etc.
        review_df["human_final_score"] = (
            review_df["human_final_score"]
            .astype(str)
            .str.extract(r"([1-3])")[0]
        )

        review_df["human_final_score"] = pd.to_numeric(
            review_df["human_final_score"],
            errors="coerce"
        )

        # Keep only completed human-review rows
        reviewed_df = review_df[
            review_df["human_final_score"].notna()
        ].copy()

        # Keep only useful human-review columns
        available_review_cols = [
            col for col in human_review_cols if col in reviewed_df.columns
        ]

        reviewed_df = reviewed_df[
            key_cols + available_review_cols
        ].copy()

        if len(reviewed_df) > 0:
            # Remove any old human-review columns from final_df before merging
            final_df = final_df.drop(
                columns=[col for col in human_review_cols if col in final_df.columns],
                errors="ignore"
            )

            final_df = final_df.merge(
                reviewed_df,
                on=key_cols,
                how="left"
            )

            # Use human score when available; otherwise keep original final score
            final_df["final_score_after_qc"] = final_df["human_final_score"].combine_first(
                final_df["final_score"]
            )

            final_df["final_label_after_qc"] = final_df["final_score_after_qc"].map(
                score_to_label
            )

            final_df["human_review_completed"] = final_df["human_final_score"].notna()

            print("Merged completed human-review results.")
            print("Human-reviewed rows:", len(reviewed_df))
        else:
            print("No completed human-review scores found.")
            print("Original results kept.")
else:
    print("No flagged_for_review.csv found.")
    print("Original results kept.")

# Save final QC dataset
final_df.to_csv(final_output_path, index=False)

print("Final output:", final_output_path.resolve())

final_df.head()
````

## Notes on the current workflow

Current intended workflow:

1. OpenAI model from `OPENAI_MODEL` / default `gpt-5.2` creates module summaries.
2. The same OpenAI model screens responses as `scorable` vs. `insufficient`.
3. The same OpenAI model generates question-specific rubrics.
4. Only `scorable` responses are sent to the three scoring models:
   - OpenAI GPT
   - AWS Bedrock Claude
   - GCP Vertex Gemini
5. The notebook uses majority vote to create `final_score`.
6. Rows with low agreement are exported for manual review.
7. Insufficient responses are kept in the final dataset and assigned `final_score = 1`.
